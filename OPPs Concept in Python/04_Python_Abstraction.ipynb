Abstraction = Hiding complexity and exposing only essential features.

In Python, abstraction is implemented using:

Abstract Base Classes (ABC)

@abstractmethod

Interfaces-like patterns

Hiding internal logic using private methods

```python
from abc import ABC, abstractmethod

# ============================================================
# CONCEPT 4 — ABSTRACTION
# ============================================================
# Abstraction means hiding INTERNAL implementation details
# and exposing ONLY essential behavior to the outside user.
#
# Python achieves abstraction primarily through:
# 1. Abstract Base Classes (ABC)
# 2. @abstractmethod decorator
# 3. Private methods (__method)
#
# The user should ONLY interact with high-level methods.
# The complex internal workings remain hidden.
# ============================================================


# ------------------------------------------------------------
# Example 1 — Basic Abstract Class
# ------------------------------------------------------------
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass  # subclasses MUST override this method

#v = Vehicle() # ERROR: Cannot instantiate abstract class

# Explanation:
# Abstract classes cannot be instantiated.
# They enforce that subclasses MUST implement certain behavior.

```

```python

# ------------------------------------------------------------
# Example 2 — Concrete subclass implementing an abstract method
# ------------------------------------------------------------
class Car(Vehicle):
    def start(self):
        print("Car started")

c = Car()
c.start()

# Explanation:
# Car MUST override start() because Vehicle declares it abstract.
# This enforces structure across related classes.
```

```python


# ------------------------------------------------------------
# Example 3 — Abstract class with multiple abstract methods
# ------------------------------------------------------------
class Shape(ABC):

    @abstractmethod
    def area(self): pass

    @abstractmethod
    def perimeter(self): pass

# Explanation:
# Any subclass of Shape must provide BOTH methods:
# area() and perimeter().

```

```python

# ------------------------------------------------------------
# Example 4 — Using abstraction for consistent API design
# ------------------------------------------------------------
class Database(ABC):

    @abstractmethod
    def connect(self): pass

    @abstractmethod
    def disconnect(self): pass

class MySQL(Database):
    def connect(self):
        print("Connected to MySQL")

    def disconnect(self):
        print("Disconnected from MySQL")

db = MySQL()
db.connect()
db.disconnect()

# Explanation:
# Abstraction ensures all databases follow the SAME method structure,
# even though each database has a different implementation.

```

```python

# ------------------------------------------------------------
# Example 5 — Abstract class with implemented (non-abstract) methods
# ------------------------------------------------------------
class Animal(ABC):

    @abstractmethod
    def sound(self): pass

    def info(self):
        print("This is an animal")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

    def info(self):
        super().info() # Correctly calls the parent Animal's info method



d = Dog()
d.sound()
d.info()

# Explanation:
# Abstract classes can contain both:
# 1. abstract methods (must be overridden)
# 2. regular methods (can be used directly)
# This allows common logic to be shared.
```

```python

# ------------------------------------------------------------
# Example 6 — Encapsulation using private helper methods
# ------------------------------------------------------------
class Engine:

    def start(self):
        # user interacts ONLY with this high-level method
        self.__fuel_check()
        self.__oil_check()
        print("Engine started successfully")

    def __fuel_check(self):
        print("Checking fuel")

    def __oil_check(self):
        print("Checking oil")

e = Engine()
e.start()
e.__fuel_check() # ERROR: Cannot access private method

# Explanation:
# __fuel_check() and __oil_check() are PRIVATE.
# User doesn't need to know how they work.
# This hides complexity — pure encapsulation.
```

```python

# ------------------------------------------------------------
# Example 7 — Abstraction using method overriding in subclasses
# ------------------------------------------------------------
class Payment(ABC):

    @abstractmethod
    def pay(self, amount): pass

class UPI(Payment):
    def pay(self, amount):
        print(f"Paid {amount} via UPI")

class Card(Payment):
    def pay(self, amount):
        print(f"Paid {amount} via Debit/Credit Card")

p1 = UPI()
p1.pay(500)

p2 = Card()
p2.pay(1000)

# Explanation:
# Different payment modes SHARE the same high-level method name.
# But each implements its own internal logic.

```

```python

# ------------------------------------------------------------
# Example 8 — Factory pattern using abstraction
# ------------------------------------------------------------
class Logger(ABC):

    @abstractmethod
    def log(self, msg): pass

class FileLogger(Logger):
    def log(self, msg):
        print(f"Writing '{msg}' to file")

class ConsoleLogger(Logger):
    def log(self, msg):
        print(f"Printing on console: {msg}")

def get_logger(type):
    if type == "file":
        return FileLogger()
    return ConsoleLogger()

logger = get_logger("console")
logger.log("Hello World")

# Explanation:
# User does NOT know which logger is created internally.
# They only interact with the abstract interface (log()).

```

```python

# ------------------------------------------------------------
# Example 9 — Abstraction using property decorators
# ------------------------------------------------------------
class Product(ABC):

    @property
    @abstractmethod
    def price(self): pass
    

class Mobile(Product):
    def __init__(self, price):
        self._price = price

    @property
    def price(self):
        return self._price

m = Mobile(25000)
print(m.price)

# Explanation:
# Abstract properties enforce rules for attributes.
# Subclasses must implement them.

```

```python

# ------------------------------------------------------------
# Example 10 — Template Method Pattern (high-level steps defined)
# ------------------------------------------------------------
class Report(ABC):

    def generate(self):
        # Template method: defines the STEPS of the algorithm
        self.header()
        self.body()
        self.footer()

    @abstractmethod
    def header(self): pass

    @abstractmethod
    def body(self): pass

    @abstractmethod
    def footer(self): pass

class SalesReport(Report):
    def header(self): print("Sales Report Header")
    def body(self): print("Sales Data Body")
    def footer(self): print("Sales Report Footer")

r = SalesReport()
r.generate()

# Explanation:
# generate() defines the ORDER of execution.
# Subclasses fill in the actual implementation.
# This is a classic abstraction pattern.
```
