Polymorphism = “many forms”
Same method name → different behaviors depending on object, context, or class type.

Forms in Python:

Method overriding

Duck typing

Built-in polymorphic functions

Operator polymorphism

Inheritance-based polymorphism

```python
# ============================================================
# CONCEPT 6 — POLYMORPHISM
# ============================================================
# Polymorphism allows the same method or operation
# to behave differently depending on the object or data type.
#
# Python supports several types of polymorphism:
# 1. Method Overriding (same method name, different implementations)
# 2. Duck Typing ("if it behaves like a duck, it's a duck")
# 3. Operator Overloading (+ works for numbers, strings, lists, etc.)
# 4. Built-in polymorphic functions (len(), max(), etc.)
# 5. Inheritance + Polymorphic behavior
# ============================================================


# ------------------------------------------------------------
# Example 1 — Method Overriding (basic polymorphism)
# ------------------------------------------------------------
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

for pet in (Dog(), Cat()):
    pet.speak()

# Explanation:
# Each class implements speak() differently.
# Loop treats all objects uniformly → polymorphic behavior.


```

```python

# ------------------------------------------------------------
# Example 2 — Duck Typing (Pythonic polymorphism)
# ------------------------------------------------------------
class Laptop:
    def code(self):
        print("Coding on Laptop")

class Mobile:
    def code(self):
        print("Coding on Mobile")

def typing_device(device):
    device.code()

typing_device(Laptop())
typing_device(Mobile())

# Explanation:
# No inheritance required.
# If an object has code() → it's valid.
# "Behavior matters more than type."
```

```python

# ------------------------------------------------------------
# Example 3 — Built-in polymorphic function: len()
# ------------------------------------------------------------
print(len("Amit"))       # string
print(len([1, 2, 3]))    # list
print(len({"a": 1}))     # dictionary

# Explanation:
# len() behaves differently based on input type.
# Function name is same, behavior is adapted → polymorphism.
```

```python


# ------------------------------------------------------------
# Example 4 — Operator Polymorphism
# ------------------------------------------------------------
print(10 + 20)        # addition
print("Amit" + " Yadav")  # string concatenation
print([1, 2] + [3, 4])    # list merging

# Explanation:
# '+' works differently for numbers, strings, and lists.
# Operator overloading enables polymorphism.

```

```python

# ------------------------------------------------------------
# Example 5 — Same method name in unrelated classes
# ------------------------------------------------------------
class Mahindra:
    def start(self):
        print("Mahindra engine start")

class Tata:
    def start(self):
        print("Tata engine start")

for car in (Mahindra(), Tata()):
    car.start()

# Explanation:
# Polymorphism does NOT require inheritance.
# As long as method name exists, Python invokes it.

```

```python

# ------------------------------------------------------------
# Example 6 — Polymorphism with Inheritance
# ------------------------------------------------------------
class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

shapes = [Square(5), Rectangle(4, 6)]

for s in shapes:
    print(s.area())

# Explanation:
# All shapes share the same interface: area()
# Actual behavior depends on concrete class.

```

```python

# ------------------------------------------------------------
# Example 7 — Polymorphism using Abstract Base Classes
# ------------------------------------------------------------
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amt): pass

class UPI(Payment):
    def pay(self, amt):
        print(f"Paid {amt} via UPI")

class Card(Payment):
    def pay(self, amt):
        print(f"Paid {amt} via Credit/Debit Card")

for method in (UPI(), Card()):
    method.pay(500)

# Explanation:
# Abstract classes enforce polymorphism.
# Subclasses must implement the abstract method.

```

```python

# ------------------------------------------------------------
# Example 8 — Polymorphism with iterables
# ------------------------------------------------------------
for item in ("Hello", [10, 20, 30], (1, 2, 3)):
    for i in item:
        print(i, end=" ")
    print()

# Explanation:
# Strings, lists, and tuples ALL support iteration.
# for-loop works polymorphically with any iterable object.


```

```python

# ------------------------------------------------------------
# Example 9 — Polymorphism using __str__ overriding
# ------------------------------------------------------------
class Person:
    def __str__(self):
        return "Person object"

class Employee:
    def __str__(self):
        return "Employee object"

print(Person())
print(Employee())

# Explanation:
# print() calls __str__() internally.
# Each class returns its own custom string representation.
```

```python

# ------------------------------------------------------------
# Example 10 — Polymorphism with function expecting generic behavior
# ------------------------------------------------------------
class FileLogger:
    def write(self, msg):
        print(f"File write: {msg}")

class ConsoleLogger:
    def write(self, msg):
        print(f"Console write: {msg}")

def log_message(logger):
    logger.write("System Error!")

log_message(FileLogger())
log_message(ConsoleLogger())

# Explanation:
# The function log_message() is POLYMORPHIC.
# It doesn't care what the logger object is.
# It only cares that write() exists.
# "Write once, work with many types."
```
