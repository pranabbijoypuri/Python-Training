```python
# ============================================================
# CONCEPT 2 — OBJECT IN PYTHON
# ============================================================
# An object is an INSTANCE of a class.
# A class defines the blueprint; the object contains the real data.
# Objects store instance attributes inside their __dict__ unless __slots__ is used.
# ============================================================


# ------------------------------------------------------------
# Example 1 — Creating simple objects
# ------------------------------------------------------------
class Student:
    def __init__(self, name):
        # 'name' becomes an instance attribute
        self.name = name

# Creating two separate objects
s1 = Student("Amit")
s2 = Student("Yadav")

# Each object holds its own data
print(s1.name)  # Amit
print(s2.name)  # Yadav

# Explanation:
# Objects store attributes in their internal dictionary.
# s1.__dict__ = {'name': 'Amit'}
# s2.__dict__ = {'name': 'Yadav'}
# Both objects share the SAME structure but hold DIFFERENT DATA.

```

```python

# ------------------------------------------------------------
# Example 2 — Object with multiple attributes
# ------------------------------------------------------------
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

c1 = Car("Mahindra", "XUV 3XO")
c2 = Car("Tata", "Punch")

print(c1.brand, c1.model)
print(c2.brand, c2.model)

# Explanation:
# Objects carry REAL VALUES for attributes defined in the class.
# Class = blueprint, object = actual filled form.

```

```python

# ------------------------------------------------------------
# Example 3 — Calling methods using objects
# ------------------------------------------------------------
class Dog:
    def bark(self):
        print("Woof!")

d = Dog()
d.bark()

# Explanation:
# 'self' inside bark() refers to the 'd' object.
# Methods are just functions that automatically receive the object as the first parameter.

```

```python

# ------------------------------------------------------------
# Example 4 — Objects can have attributes added dynamically
# ------------------------------------------------------------
class Demo:
    pass   # empty class

obj = Demo()

# Adding attributes at runtime
obj.x = 100
obj.y = "Dynamic Attribute"

print(obj.x, obj.y)

# Explanation:
# Python's dynamic object model allows adding attributes to any object.
# Internally it updates obj.__dict__ with new entries:
# obj.__dict__ = {'x': 100, 'y': 'Dynamic Attribute'}

```

```python

# ------------------------------------------------------------
# Example 5 — Passing objects to functions
# ------------------------------------------------------------
class Person:
    def __init__(self, name):
        self.name = name

def greet(p):
    # 'p' is an object, so we can access its attributes
    print("Hello", p.name)


greet(Person("Amit"))

# Explanation:
# Objects can be passed to functions like variables.
# Functions then access properties of the object.

```

```python

# ------------------------------------------------------------
# Example 6 — Returning objects from functions
# ------------------------------------------------------------
class Laptop:
    def __init__(self, brand):
        self.brand = brand

def create_laptop():
    return Laptop("Dell")  # returning a new object

lp = create_laptop()
print(lp.brand)

# Explanation:
# Functions can build and return objects.
# Useful for factory patterns, alternate constructors, etc.

```

```python

# ------------------------------------------------------------
# Example 7 — One class, multiple objects (Instance independence)
# ------------------------------------------------------------
class Counter:
    def __init__(self):
        self.count = 0

c1 = Counter()
c2 = Counter()

c1.count += 10
c2.count += 5

print(c1.count, c2.count)

# Explanation:
# Each object has its own 'count'.
# Changing one object’s attribute does NOT affect another.
# This demonstrates object-level state.

```

```python

# ------------------------------------------------------------
# Example 8 — Objects referencing other objects (Composition)
# ------------------------------------------------------------
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()  # Car contains Engine object

c = Car()
c.engine.start()

# Explanation:
# Objects can contain OTHER objects.
# This is called Composition — “has-a” relationship.
# Car HAS-A Engine.

```

```python


# ------------------------------------------------------------
# Example 9 — Printing the object (Built-in __dict__)
# ------------------------------------------------------------
class Box:
    def __init__(self, w, h):
        self.width = w
        self.height = h

b = Box(10, 5)
print(b.__dict__)

# Explanation:
# __dict__ shows all instance attributes stored inside the object.
# Output: {'width': 10, 'height': 5}
# Very useful for debugging.
```

```python

# ------------------------------------------------------------
# Example 10 — Objects with same methods but independent data
# ------------------------------------------------------------
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amt):
        self.balance += amt

acc1 = BankAccount(1000)
acc2 = BankAccount(500)

acc1.deposit(200)
acc2.deposit(300)

print(acc1.balance)  # 1200
print(acc2.balance)  # 800

# Explanation:
# acc1 and acc2 both use the SAME deposit() method
# BUT they modify DIFFERENT internal balances.
# This is one of the core strengths of OOP: shared behavior, separate data.
```
