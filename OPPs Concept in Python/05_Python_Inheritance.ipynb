Inheritance allows a class (child/subclass) to acquire:

attributes

methods

from another class (parent/superclass).

This enables:

code reuse

method overriding

hierarchical design

polymorphism

```python
# ============================================================
# CONCEPT 5 — INHERITANCE
# ============================================================
# Inheritance = "A child class acquiring features from parent class"
#
# Types of Inheritance in Python:
# 1. Single
# 2. Multilevel
# 3. Hierarchical
# 4. Multiple
# 5. Hybrid (combination)
#
# Child class can:
# - reuse parent methods
# - override parent methods
# - extend parent functionality
# ============================================================


# ------------------------------------------------------------
# Example 1 — Single Inheritance (basic)
# ------------------------------------------------------------
class Parent:
    def show(self):
        print("This is Parent class")

class Child(Parent):
    pass

c = Child()
c.show()

# Explanation:
# Child inherits show() from Parent.
# No need to rewrite the method — classic code reuse.

```

```python

# ------------------------------------------------------------
# Example 2 — Overriding a Parent Method
# ------------------------------------------------------------
class Car:
    def start(self):
        print("Car starting...")

class XUV(Car):
    def start(self):
        print("XUV starting with turbo engine...")

x = XUV()
x.start()

# Explanation:
# Child provides its OWN version of start().
# This allows more specific behavior.

```

```python

# ------------------------------------------------------------
# Example 3 — Calling Parent Method from Child (super keyword)
# ------------------------------------------------------------
class Animal:
    def sound(self):
        print("Generic animal sound")

class Dog(Animal):
    def sound(self):
        super().sound()        # calling parent method
        print("Dog barks")

d = Dog()
d.sound()

# Explanation:
# super() allows calling parent version FIRST,
# then child adds additional behavior.

```

```python

# ------------------------------------------------------------
# Example 4 — Multilevel Inheritance (A → B → C)
# ------------------------------------------------------------
class A:
    def showA(self):
        print("This is A")

class B(A):
    def showB(self):
        print("This is B")

class C(B):
    def showC(self):
        print("This is C")

obj = C()
obj.showA()
obj.showB()
obj.showC()

# Explanation:
# Class C inherits from B, which inherits from A.
# So C gets features of BOTH A and B.

```

```python

# ------------------------------------------------------------
# Example 5 — Hierarchical Inheritance (one parent, many children)
# ------------------------------------------------------------
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal): pass
class Cat(Animal): pass

Dog().speak()
Cat().speak()

# Explanation:
# Both Dog and Cat inherit from Animal.
# Same parent → multiple children.

```

```python

# ------------------------------------------------------------
# Example 6 — Multiple Inheritance (two parents)
# ------------------------------------------------------------
class Father:
    def skills(self):
        print("Father: Coding")

class Mother:
    def skills(self):
        print("Mother: Cooking")

class Child(Father, Mother):
    pass

child = Child()
child.skills()

# Explanation:
# Python follows MRO (Method Resolution Order).
# Father comes first in inheritance list,
# so Father.skills() is executed.
# Can be changed using class ordering.

```

```python

# ------------------------------------------------------------
# Example 7 — Using super() with Multiple Inheritance
# ------------------------------------------------------------
class A:
    def show(self):
        print("A show")

class B(A):
    def show(self):
        print("B show")
        super().show()

class C(B):
    def show(self):
        print("C show")
        super().show()

obj = C()
obj.show()

# Explanation:
# super() follows MRO, not just parent class.
# MRO for C: C → B → A.
# Output prints in that chain.

```

```python

# ------------------------------------------------------------
# Example 8 — Constructor Inheritance (super in __init__)
# ------------------------------------------------------------
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)  # call parent constructor
        self.salary = salary

e = Employee("Amit", 50000)
print(e.name, e.salary)

# Explanation:
# Child class needs to call parent constructor explicitly.
# super() ensures attributes from parent are initialized correctly.

```

```python

# ------------------------------------------------------------
# Example 9 — Method extension (not overriding fully)
# ------------------------------------------------------------
class Bike:
    def start(self):
        print("Bike starting...")

class SportsBike(Bike):
    def start(self):
        print("Checking safety systems...")
        super().start()

s = SportsBike()
s.start()

# Explanation:
# Child adds new steps BEFORE calling parent method.

```

```python

# ------------------------------------------------------------
# Example 10 — Hybrid Inheritance (combination of types)
# ------------------------------------------------------------
class A:
    def a(self):
        print("A")

class B(A):
    def b(self):
        print("B")

class C(A):
    def c(self):
        print("C")

class D(B, C):
    def d(self):
        print("D")

obj = D()
obj.a()
obj.b()
obj.c()
obj.d()

# Explanation:
# Hybrid = mixture of multiple inheritance styles.
# MRO ensures correct method lookup order.
# D inherits from B and C, both inherit from A → diamond structure.
```
