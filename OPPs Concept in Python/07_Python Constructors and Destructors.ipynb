Python provides:

Constructor → __init__()
Automatically runs when an object is created.

Destructor → __del__()
Automatically runs when an object is destroyed (usually when reference count becomes zero or program ends).

```python
# ============================================================
# CONCEPT 7 — CONSTRUCTORS & DESTRUCTORS
# ============================================================
# Constructor: __init__()
# - Initializes object attributes
# - Runs automatically when an object is created
#
# Destructor: __del__()
# - Called when object is about to be destroyed
# - Useful for cleanup (files, DB connections, logs)
#
# NOTE: Python's garbage collector decides when __del__ executes.
# ============================================================


# ------------------------------------------------------------
# Example 1 — Basic constructor
# ------------------------------------------------------------
class Person:
    def __init__(self):
        # Constructor initializing default data
        self.name = "Unnamed"

p = Person()
print(p.name)

# Explanation:
# __init__ runs instantly when Person() is called.
# It initializes 'name' for the new object.

```

```python

# ------------------------------------------------------------
# Example 2 — Constructor with parameters
# ------------------------------------------------------------
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s = Student("Amit", 95)
print(s.name, s.marks)

# Explanation:
# Constructor receives arguments required to build the object.

```

```python

# ------------------------------------------------------------
# Example 3 — Constructor performing calculations
# ------------------------------------------------------------
class Rectangle:
    def __init__(self, w, h):
        self.area = w * h  # computing during initialization

r = Rectangle(4, 5)
print(r.area)

# Explanation:
# Constructors can compute values that the object must store.

```

```python

# ------------------------------------------------------------
# Example 4 — Calling methods inside constructor
# ------------------------------------------------------------
class Car:
    def __init__(self):
        print("Creating Car...")
        self.start_engine()

    def start_engine(self):
        print("Engine started")

c = Car()

# Explanation:
# Constructors often call helper methods to set up object state.

```

```python

# ------------------------------------------------------------
# Example 5 — Multiple objects initialized differently
# ------------------------------------------------------------
class Laptop:
    def __init__(self, brand, ram):
        self.brand = brand
        self.ram = ram

l1 = Laptop("HP", "8GB")
l2 = Laptop("Dell", "16GB")

print(l1.brand, l1.ram)
print(l2.brand, l2.ram)

# Explanation:
# Each object is initialized independently with different values.

```

```python

# ------------------------------------------------------------
# Example 6 — Default values in constructor
# ------------------------------------------------------------
class Account:
    def __init__(self, balance=0):
        self.balance = balance

a1 = Account()
a2 = Account(1000)

print(a1.balance)
print(a2.balance)

# Explanation:
# Default constructor values make arguments optional.
```

```python


# ------------------------------------------------------------
# Example 7 — Destructor (basic usage)
# ------------------------------------------------------------
class Demo:
    def __del__(self):
        print("Destructor called. Object removed.")

d = Demo()
del d  # manual deletion

# Explanation:
# __del__ runs when object is deleted or garbage-collected.
# Python decides final timing, but del forces early cleanup.
```

```python


# ------------------------------------------------------------
# Example 8 — Destructor for file handling
# ------------------------------------------------------------
class FileHandler:
    def __init__(self, filename):
        self.file = open(filename, "w")
        print("File opened")

    def write(self, text):
        self.file.write(text)

    def __del__(self):
        self.file.close()
        print("File closed")

fh = FileHandler("demo.txt")
fh.write("Hello!")

del fh  # triggers destructor

# Explanation:
# __del__ ensures file is properly closed even if user forgets.
# Important for resource management.
```

```python


# ------------------------------------------------------------
# Example 9 — Constructor + Destructor (lifecycle demonstration)
# ------------------------------------------------------------
class Test:
    def __init__(self):
        print("Object Created")

    def __del__(self):
        print("Object Destroyed")

t = Test()
t2 = Test()

del t

# Explanation:
# Constructor messages appear when objects are created.
# Destructor messages appear when objects are destroyed.

```

```python

# ------------------------------------------------------------
# Example 10 — Tracking active objects with constructors/destructors
# ------------------------------------------------------------
class Counter:
    active_objects = 0

    def __init__(self):
        Counter.active_objects += 1
        print("Object created. Active:", Counter.active_objects)

    def __del__(self):
        Counter.active_objects -= 1
        print("Object destroyed. Active:", Counter.active_objects)

c1 = Counter()
c2 = Counter()

del c1

# Explanation:
# Constructor increments active count.
# Destructor decrements count.
# Demonstrates object lifecycle tracking.
```
