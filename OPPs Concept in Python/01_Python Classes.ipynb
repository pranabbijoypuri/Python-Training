📘 CONCEPT 1 — CLASS in Python (10 Detailed Examples with Deep Explanation)

A class is a blueprint for creating objects.
It defines:

Attributes (data)

Methods (functions/behavior)

🔵 Example 1 — Basic Class with a Class Attribute
✅ Purpose:

Demonstrates how to create a simple class with a shared attribute.

```python
class Car:
    wheels = 4   # class attribute shared by all objects

print(Car.wheels)
```

```python
car = Car()
car.wheels
```



📝 Explanation:

wheels belongs to the class, not to any specific object.

All Car objects will have wheels = 4 unless overridden.

You can access it using Car.wheels or obj.wheels.


🔵 Example 2 — Class with Instance Attributes (__init__)
✅ Purpose:

```python

#Shows how to define attributes unique to each object.

class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

c = Car("Mahindra", "XUV 3XO")
print(c.brand, c.model)
```



📝 Explanation:

__init__ is the constructor; it runs automatically when you create an object.

self.brand and self.model belong to the object, not the class.

Each object stores its own data.

🔵 Example 3 — Class with Methods
✅ Purpose:

Shows behavior defined inside a class.

```python

class Calculator:
    def add(self, a, b):
        return a + b

calc = Calculator()
print(calc.add(10, 20))
 # Uses parameters only --> no state stored in the object
```

```python
class Calculator:
  def add(self, a, b):
    self.a = a
    self.b= b
    return self.a + self.b

calc = Calculator()
print(calc.add(10, 20))
# Copies parameters into self--> updates the objects'state.
```

```python
calc.a
```

```python
calc.b
```

📝 Explanation:

A method is a function defined inside a class.

The first parameter self refers to the object calling the method.


🔵 Example 4 — Using Static Methods (@staticmethod)
✅ Purpose:

Shows methods that do NOT depend on object or class state.

```python

class Math:
    @staticmethod
    def square(x):
        return x * x

print(Math.square(6))
```

📝 Explanation:

Static methods behave like regular functions but are placed inside a class for grouping.

They do not use self or cls.




🔵 Example 5 — Using Class Methods (@classmethod)
❗ This is important and often asked in interviews.
✅ Purpose:

Shows how to access and modify class-level data using cls.


```python

class Employee:
    company = "Infosys"  # class-level attribute

    @classmethod
    def show_company(clss):
        print("Company:", clss.company)

Employee.show_company()
```

```python
class Employee:
    company = "Infosys"  # class-level attribute

    @classmethod
    def show_company(clss):
      e1 = Employee()
      e2= Employee()

      print("Company:", clss.company)
      print("Company:", e1.company)


Employee.show_company()
Employee.company
```

📝 Deep Explanation (Clear & Simple):
What is a Class Method?

A method that receives the class itself as the first parameter instead of the object.

Type	First parameter	Access to	Use case
Instance method	self	Object data	Default behavior
Class method	cls	Class data	Shared state
Static method	No self or cls	Utility functions	Independent logic
✔ How it works:

When you call Employee.show_company(), Python passes the class (Employee) automatically as cls.

So inside the method:

cls.company == Employee.company


This allows:

Reading class attributes

Modifying class attributes

Creating alternate constructors

✔ Why use a class method?

Because sometimes you want to modify or access class-level shared data rather than object-specific data.

Example:

Changing company name for ALL employees

Creating objects from alternative formats (e.g., from a string)

🔵 Example 6 — Empty Class + Dynamic Attribute Assignment
❗ This confuses many beginners — here’s the simplest explanation.

```python

class Demo:
    pass # do nothing

obj = Demo()
obj.name = "Amit"
obj.age = 30
print(obj.name, obj.age)
```

```python

obj1 = Demo()
obj1.namee = "Amit"
obj1.agee = 30
print(obj1.namee, obj1.agee)
```



📝 Deep Explanation:
✔ What is happening?

Python allows objects to have attributes added at runtime, even if the class doesn't define them.

The Demo class is empty — BUT Python objects have a flexible internal structure that allows:

obj.new_attribute = value


This is called dynamic attribute binding.

✔ Why does Python allow this?

Because Python uses a dynamic object model.

Every object has an internal dictionary called:

obj.__dict__


When you do:

obj.name = "Amit"


Internally Python does:

obj.__dict__["name"] = "Amit"


Which means:

You can attach ANY data to ANY object

The class doesn’t need predefined attributes

✔ When is this useful?

Creating flexible objects

Prototyping

Mock objects in testing

Quick data containers (similar to structs in C)

✔ Important Note:

This works only if the class does not use __slots__.


🔵 Example 7 — Private Class Attribute

```python


class Bank:
    interest_rate = 7.5  # private class attribute



Bank.interest_rate
```

```python
class Bank:
    __interest_rate = 7.5  # private class attribute
    def __init__(self):
        self.__interest_rate = 8.5


    @classmethod
    def show_rate(cls):
        print("Interest Rate (Class):", cls.__interest_rate)

Bank.show_rate()
b= Bank()
# Accessing the name-mangled instance attribute
print("Interest Rate (Instance):", b._Bank__interest_rate)
```

```python
Bank.__interest_rate # Cannot be used outside the class,
```




📝 Explanation:

__interest_rate is private to the class.

Only cls.__interest_rate inside the class can access it.


🔵 Example 8 — Class with Multiple Methods

```python
#Optimized Code.
class Student:
    def __init__(self, name):
        self.name = name

    def study(self):
        print(self.name, "is studying")

    def sleep(self):
        print(self.name, "is sleeping")

s = Student("Amit")
s.study()
s.sleep()
```

```python
# Unoptimized code
class Student:

    def study(self, name):
        print(name, "is studying")

    def sleep(self, name):
        print(name, "is sleeping")

s = Student()
s.study('Amit')
s.sleep('Amit')
```




📝 Explanation:

Shows how multiple behaviors belong to one blueprint.


🔵 Example 9 — Class with Default Parameter Values

```python
class Laptop:
    def __init__(self, brand="HP", ram="8GB"):
        self.brand = brand
        self.ram = ram

lp = Laptop()
print(lp.brand, lp.ram)

lp1 = Laptop('Dell')
print(lp1.brand, lp1.ram)

lp2 = Laptop('Levevo', '16GB')
print(lp2.brand, lp2.ram)
```


📝 Explanation:

If no arguments are provided, default values are used.

Objects can still override them.

```python
class Engine:
    def __init__(self):
        self.status = "OFF"

    def __start_engine(self):   # private helper
        self.status = "ON"

    def start(self):
        self.__start_engine()
        print("Engine Started! Status =", self.status)

e = Engine()
e.start()
```



📝 Explanation:

__start_engine is private: used only inside the class.

It hides internal logic from the outside world.
