# Missing Data

Let's show a few convenient methods to deal with Missing Data in pandas:

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'A':[1,2,np.nan],
                  'B':[5,np.nan,np.nan],
                  'C':[1,2,3]})
```

```python
df
```

```python
df.dropna() #inplace=True not done so its not getting effective in the original DF.
# axis=0 by default.
```

```python
df
```

```python
df.dropna(axis=1) # Drop all the columns where you find any null value.
```

```python
df.dropna(thresh=2) # Atleast 1 non null value per row.
```

```python
df.fillna(value='FILL VALUE')
```

```python
df['A'].fillna(value=df['A'].mean())
```

```python
df.fillna(df.mean()) # fill up all the columns with their mean value(mean value of the column). Only for numeric columns.
```
