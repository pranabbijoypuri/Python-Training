## Time Series in Pandas

```python
#Here we will explore the DateTime functions provided by Pandas and how efficient it is to analyse TIme Series Data.
import pandas as pd
air_df = pd.read_csv('https://raw.githubusercontent.com/amity1415/eKeeda/refs/heads/main/Rawdata/AQI.csv')
air_df.head()
```

```python
# Date and Time columns are object datatype. We will need to convert them to datetime format.
air_df.info()
```

```python
air_df['Date'] = pd.to_datetime(air_df['Date'], format='%m/%d/%Y')
air_df.info()
```

```python
air_df['Time'].unique()
```

```python
air_df['Time'] = pd.to_datetime(air_df['Time'], format='%H:%M:%S')
air_df.info()
```

```python
air_df.head()
```

```python
# Extrace year from the Date Column
year = air_df['Date'].dt.year
year
air_df['Year']= year
```

```python
air_df.head()
```

```python
#Move the Year column from last to 1st.
last_col = air_df.pop('Year')
air_df.insert(0, 'year', last_col)
air_df.head()
```

```python
#Extract Month from the Date Column
month = air_df['Date'].dt.month
month
air_df.insert(1, 'month', month)
air_df.head()
```

```python
month = air_df['Date'].dt.month_name()
month
air_df.insert(2, 'month_name', month)
air_df.head()
```

```python
day = air_df.Date.dt.day
day
air_df.insert(3, 'day', day)
air_df.head()
```

```python
# Extract the Week number from Date Column
week = air_df.Date.dt.weekday_name
week
air_df.insert(5, 'weekday_name', week)
air_df.head()
```

```python
#Print the Week Number from the date
week_num = air_df.Date.dt.isocalendar().week
week_num
air_df.insert(4, 'week_num', week_num)
air_df.head()
```

```python
#https://overapi.com/python --> Use this link for more Python Operations
```

```python
day_of_year = air_df.Date.dt.day_of_year
air_df.insert(4, 'day_of_year', day_of_year)
air_df.head()
```

```python
hour = air_df.Time.dt.hour # Similarly it extracts , dt.minute , dt.second
print(hour.head())
```

Performing Conditional Operations on Time.
Lets measure the number of records before 9 am.

```python
timestamp = pd.to_datetime("09:00:00", format='%H:%M:%S')
timestamp
```

```python
air_df[air_df['Time'] < timestamp]['RH'].mean()
```

```python
air_df[air_df['Time'] > timestamp]['RH'].mean()
```

```python
#TimeDelta
from datetime import timedelta
print(air_df['Date'])
print('----------')
print(air_df['Date'] + timedelta(days=1)) # days= -1, hours, minutes, second, month, week, year.
```

```python
#Lets measure the number of records before 01/01/2005
date_stamp = pd.to_datetime("01/01/2005", format='%d/%m/%Y')
date_stamp
```

```python
air_df[air_df['Date'] < date_stamp].tail()
```
